```python
from collections import deque

def water_jug():
    q = deque([((0, 0), [])])
    visited = set()

    while q:
        (a, b), path = q.popleft()

        if (a, b) in visited:
            continue
        visited.add((a, b))

        path = path + [(a, b)]

        if a == 2:
            return path

        states = [
            (4, b),                         # Fill 4-gallon jug
            (a, 3),                         # Fill 3-gallon jug
            (0, b),                         # Empty 4-gallon jug
            (a, 0),                         # Empty 3-gallon jug
            (a - min(a, 3-b), b + min(a, 3-b)),  # 4 -> 3
            (a + min(b, 4-a), b - min(b, 4-a))   # 3 -> 4
        ]

        for state in states:
            if state not in visited:
                q.append((state, path))

solution = water_jug()

print("Solution:")
for state in solution:
    print(state)
```
