```python
from collections import deque

def solve(start):
    goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)
    q = deque([(start, [])])
    visited = {start}

    while q:
        state, path = q.popleft()

        if state == goal:
            return path + [state]

        z = state.index(0)
        r, c = divmod(z, 3)

        for dr, dc in [(1,0), (-1,0), (0,1), (0,-1)]:
            nr, nc = r + dr, c + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                nz = nr * 3 + nc
                new = list(state)
                new[z], new[nz] = new[nz], new[z]
                new = tuple(new)

                if new not in visited:
                    visited.add(new)
                    q.append((new, path + [state]))

    return None


# Input
start = tuple(map(int, input("Enter 9 numbers (0 for blank): ").split()))

solution = solve(start)

# Output
if solution:
    print("\nSolution:")
    for s in solution:
        print(s[0:3])
        print(s[3:6])
        print(s[6:9])
        print()
else:
    print("No solution")
```
