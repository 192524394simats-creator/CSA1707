```python
def is_safe(board, row, col):
    for i in range(row):
        if board[i] == col or abs(board[i] - col) == abs(i - row):
            return False
    return True


def solve(board, row):
    if row == 8:
        return True

    for col in range(8):
        if is_safe(board, row, col):
            board[row] = col

            if solve(board, row + 1):
                return True

            board[row] = -1

    return False


# Input
board = [-1] * 8

if solve(board, 0):
    print("Solution:")
    for row in range(8):
        for col in range(8):
            print("Q" if board[row] == col else ".", end=" ")
        print()
else:
    print("No solution")
```
